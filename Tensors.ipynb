# Important

## . Inside a computer, all data is stored as a flat list of numbers.

## . Matrices and tensors do NOT change the numbers.

## . They only define how those numbers are interpreted.

## . A matrix is a 2D interpretation of numbers (rows × columns).

## . A tensor is a general interpretation that supports any number of dimensions.

## . You can represent higher dimensions using matrices, but the meaning of dimensions is not stored.

## . A tensor explicitly stores shape and dimension structure, so operations know what each axis means.

## . A matrix is a special case of a tensor (2D tensor).

## . Tensors exist to preserve structure, avoid ambiguity, and automate operations in ML.

```python
weight = torch.tensor([0.1, 0.2, 0.3, 0.2, 0.2])
```
