# Torch Squeeze

## In here we can drop and add dimentions according to their size.

## "squeezing" is specifically designed to remove "empty" dimensions.

## Global Squeeze: tensor.squeeze() removes all dimensions of size 1.

## Targeted Squeeze: tensor.squeeze(dim=n) only removes the dimension at index n, and only if its size is 1.

```python
Activity:
```

```python
import torch
```

```python
x = torch.tensor([[[1,2,3], [5,6,7]]])
x
```

```python
x.shape
```

```python
x.ndim  # Number of dimentions
```

```python
torch.squeeze(x)
```

```python
torch.squeeze(x, dim=0)  # Remove the dimention 1 if it's size is 1
```

```python
y = torch.tensor([[[1],[2],[3]], [[1],[2],[3]]])
y
```

```python
y.ndim
```

```python
torch.squeeze(y)
```

```python
# torch.squeeze(y) === torch.squeeze(y, dim=2)
```

# Torch Unsqueeze

### Returns a new tensor with a dimension of size one inserted at the specified position.

```python
n = torch.tensor([[1,2,3],[5,6,7]])
```

```python
n
```

```python
n.ndim
```

```python
torch.unsqueeze(n, dim=2)
```

Operation	    Target Dimension Size	     Result			
squeeze()	    Must be 1	                 Removes the dimension; data stays the same.			
flatten()	    Any size	                 Collapses multiple dimensions into one.			
mean() / sum()	Any size	                 Reduces the dimension to size 1 (by doing math).			
reshape()	    Any size	                 Manually changes shape (Total elements must match).
